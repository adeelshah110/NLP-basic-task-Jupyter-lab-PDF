# NLP Part 4: Normalization

## What is Normalization?

**Normalization** is the process of converting text into a **standard and consistent form**.

### Example:
**HELLO! → hello**

Common normalization techniques include:
- Lowercasing
- Stemming
- Lemmatization

---

## 1. Stemming

**Stemming** removes prefixes or suffixes from a word to produce a **root-like form**.

### Examples:

| Original Word | Stem |
|---|---|
| playing | play |
| played | play |
| studies | studi |
| connected | connect |

**Important:** The stem may **not be a real dictionary word**.

Example:

**studies → studi**

---

## 2. Lemmatization

**Lemmatization** converts a word into its **correct dictionary base form**, called a **lemma**.

### Examples:

| Original Word | Lemma |
|---|---|
| playing | play |
| studies | study |
| running | run |
| better | good |

---

## Stemming vs. Lemmatization

| Stemming | Lemmatization |
|---|---|
| Cuts word endings | Finds the correct base word |
| Faster | Usually slower |
| May produce invalid words | Produces meaningful dictionary words |
| Less linguistically accurate | More linguistically accurate |
| studies → studi | studies → study |

---

## Which One is Better?

**Lemmatization** is generally better when we want **accurate and meaningful words**.

**Stemming** can be useful when we want **simple and fast text processing**.

### Easy way to remember:

> **Stemming cuts the word; Lemmatization understands the word.**

```python
import string

text = "Hello! I LOVE Natural Language Processing."

# Convert to lowercase
text = text.lower()

# Remove punctuation
text = text.translate(str.maketrans('', '', string.punctuation))

print(text)
```

## Stemming 2

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connected",
    "connection"
]

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

stemmer = PorterStemmer()

text = "The students are studying and playing with connected computers."

tokens = word_tokenize(text)

for word in tokens:
    print(word, "->", stemmer.stem(word))
```

## Lemmatization (3)

```python
from nltk.stem import WordNetLemmatizer

#nltk.download('wordnet')
#nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = [
    "playing",
    "played",
    "studying",
    "studies",
    "running"
]

for word in words:
    lemma = lemmatizer.lemmatize(word, pos='v')
    print(word, "->", lemma)
```

```python


stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["studies", "studying", "playing", "connections"]

print("WORD\t\tSTEM\t\tLEMMA")
print("-" * 40)

for word in words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word, pos='v')

    print(word, "\t", stem, "\t\t", lemma)
```
