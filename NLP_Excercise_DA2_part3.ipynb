# NLP Part 5: WordNet

## What is WordNet?

**WordNet** is a large lexical database of the English language.

It groups words with similar meanings into sets called **Synsets (Synonym Sets)**.

WordNet also stores relationships between words and concepts.

### Example

**car → automobile → motorcar**

These words have similar meanings and may belong to the same synset.

---

## Why do we use WordNet in NLP?

WordNet can be used for:

- Finding meanings of words
- Finding synonyms
- Finding antonyms
- Finding relationships between words
- Finding more general or specific concepts
- Measuring semantic similarity
- Word sense disambiguation
- Text classification and clustering

```python
import nltk

#nltk.download('wordnet')
#nltk.download('omw-1.4')

from nltk.corpus import wordnet as wn

print("WordNet is ready!")
```

# 1. Synsets

## What is a Synset?

A **Synset** means a **Synonym Set**.

It is a group of words that represent the same or a very similar concept.

A word can have **more than one synset** because the same word can have different meanings.

### Example

The word **car** can have different meanings.

WordNet stores these different meanings separately.

```python
from nltk.corpus import wordnet as wn

word = "car"

synsets = wn.synsets(word)

print("Synsets of", word)

for synset in synsets:
    print(synset)
```

# 2. Word Definitions

Each synset has its own **definition**.

This helps us understand the different meanings of a word.

### Example

A word such as **bank** can mean:

- A financial institution
- The side of a river

WordNet stores these meanings in different synsets.

```python
word = "bank"

for synset in wn.synsets(word):
    print("Synset:", synset.name())
    print("Definition:", synset.definition())
    print()
```

# 3. Synonyms

## What is a Synonym?

A **synonym** is a word that has the same or a similar meaning as another word.

### Examples

- car → automobile
- happy → glad
- big → large

WordNet allows us to find synonyms automatically.

```python
word = "happy"

synonyms = []

for synset in wn.synsets(word):
    for lemma in synset.lemmas():
        synonyms.append(lemma.name())

print("Word:", word)
print("Synonyms:")

print(set(synonyms))
```

# 4. Antonyms

## What is an Antonym?

An **antonym** is a word with the opposite meaning.

### Examples

- good ↔ bad
- happy ↔ unhappy
- hot ↔ cold
- large ↔ small

WordNet stores antonym relationships for many words.

```python
word = "good"

antonyms = []

for synset in wn.synsets(word):
    for lemma in synset.lemmas():

        if lemma.antonyms():
            antonyms.append(
                lemma.antonyms()[0].name()
            )

print("Word:", word)
print("Antonyms:", set(antonyms))
```

# 5. Synonyms and Antonyms Together

We can combine both operations.

For one word, we can find:

1. Synonyms
2. Antonyms

```python
word = "good"

synonyms = set()
antonyms = set()

for synset in wn.synsets(word):

    for lemma in synset.lemmas():

        # Add synonym
        synonyms.add(lemma.name())

        # Add antonym
        if lemma.antonyms():
            antonyms.add(
                lemma.antonyms()[0].name()
            )

print("Original Word:", word)

print("\nSynonyms:")
print(synonyms)

print("\nAntonyms:")
print(antonyms)
```

# 6. Semantic Relationships in WordNet

WordNet does more than find synonyms.

It also describes relationships between concepts.

Important relationships include:

| Relationship | Meaning | Example |
|---|---|---|
| Synonym | Similar meaning | car → automobile |
| Antonym | Opposite meaning | good → bad |
| Hypernym | More general concept | car → vehicle |
| Hyponym | More specific concept | car → sports car |
| Meronym | Part of something | car → wheel |

# 7. Hypernyms

## What is a Hypernym?

A **hypernym** is a more general concept.

### Example

**car → vehicle**

A car is a type of vehicle.

Therefore:

**vehicle is a hypernym of car.**

Another example:

**dog → animal**

```python
car = wn.synset('car.n.01')

print("Word:")
print(car)

print("\nDefinition:")
print(car.definition())

print("\nHypernyms:")
print(car.hypernyms())
```

```python
car = wn.synset('car.n.01')

for hypernym in car.hypernyms():
    print(hypernym.name())
    print(hypernym.definition())
```

# 8. Hyponyms

## What is a Hyponym?

A **hyponym** is a more specific concept.

For example:

**vehicle**
↓
**car**
↓
**sports car**

A sports car is a more specific type of car.

Therefore:

**sports car is a hyponym of car.**

```python
car = wn.synset('car.n.01')

hyponyms = car.hyponyms()

print("Hyponyms of car:")

for word in hyponyms:
    print(word.name())
```

# 9. Meronyms

## What is a Meronym?

A **meronym** represents a part of something.

### Example

A car contains:

- wheel
- engine
- door

Therefore, a **wheel is part of a car**.

This is called a **part-of relationship**.

```python
car = wn.synset('car.n.01')

parts = car.part_meronyms()

print("Parts of a car:")

for part in parts:
    print(part.name())
```

# 10. Complete WordNet Example

Now we will combine several WordNet operations.

For a given word, we will display:

1. Synsets
2. Definition
3. Synonyms
4. Hypernyms
5. Hyponyms

```python
from nltk.corpus import wordnet as wn

word = "dog"

synsets = wn.synsets(word)

print("WORD:", word)
print("=" * 50)

# Show available synsets
print("\nSYNSETS:")

for synset in synsets:
    print(synset.name())

# Select first meaning
first_synset = synsets[0]

print("\nSELECTED SYNSET:")
print(first_synset.name())

print("\nDEFINITION:")
print(first_synset.definition())

print("\nSYNONYMS:")

for lemma in first_synset.lemmas():
    print(lemma.name())

print("\nHYPERNYMS:")

for hypernym in first_synset.hypernyms():
    print(hypernym.name())

print("\nHYPONYMS:")

for hyponym in first_synset.hyponyms():
    print(hyponym.name())
```
