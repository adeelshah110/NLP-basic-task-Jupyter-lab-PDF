# NLP Part 6: Named Entity Recognition (NER)

## What is Named Entity Recognition?

**Named Entity Recognition (NER)** is an NLP technique used to identify and classify
important entities in text.

### Common Entity Types

| Entity | Meaning | Example |
|---|---|---|
| PERSON | Person's name | Steve Jobs |
| ORGANIZATION | Company or organization | Apple |
| LOCATION | Place or location | Finland |
| DATE | Date or year | 2026 |
| TIME | Time | 10:30 AM |
| MONEY | Monetary value | $500 |

### Example

**"Apple was founded by Steve Jobs in California."**

NER can identify:

- **Apple** → ORGANIZATION
- **Steve Jobs** → PERSON
- **California** → LOCATION

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

print("NLTK is ready!")
```

# 1. Basic NER Example

NER normally works after:

**Text → Tokenization → POS Tagging → Named Entity Recognition**

First, we divide the sentence into tokens.

Then we assign POS tags.

Finally, NER identifies named entities.

```python
from nltk import word_tokenize, pos_tag, ne_chunk

text = "Barack Obama visited Finland."

# Tokenization
tokens = word_tokenize(text)

# POS Tagging
pos_tags = pos_tag(tokens)

# Named Entity Recognition
entities = ne_chunk(pos_tags)

print(entities)
```

# 2. Extract Named Entities

Instead of displaying the complete NER tree,
we can extract only the named entities.

For each entity, we want to display:

**Entity Name → Entity Type**

```python
from nltk import word_tokenize, pos_tag, ne_chunk

text = """
Barack Obama visited Finland and met representatives
from Microsoft in Helsinki.
"""

tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)
tree = ne_chunk(pos_tags)

for chunk in tree:

    if hasattr(chunk, 'label'):

        entity = " ".join(
            word for word, tag in chunk.leaves()
        )

        print(entity, "->", chunk.label())
```

# What is GPE?

NLTK often uses the label **GPE**.

**GPE = Geopolitical Entity**

It represents geographical areas that also have political or administrative meaning.

Examples:

- Finland → GPE
- Pakistan → GPE
- Helsinki → GPE
- London → GPE

# 3. NER with More Text

Let's use a larger example containing:

- People
- Organizations
- Countries
- Cities

```python
text = """
Elon Musk founded SpaceX in the United States.
microsoft was founded by Bill Gates and Paul Allen.
nokia is a Finnish company headquartered in Espoo, Finland.
"""

tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)
tree = ne_chunk(pos_tags)

for chunk in tree:

    if hasattr(chunk, 'label'):

        entity = " ".join(
            word for word, tag in chunk.leaves()
        )

        print(entity, "->", chunk.label())
```

# NER Pipeline

The basic NLTK NER pipeline is:

Raw Text
↓
Tokenization
↓
POS Tagging
↓
NER
↓
Named Entities

### Example

"Steve Jobs founded Apple."

↓

Tokens

["Steve", "Jobs", "founded", "Apple", "."]

↓

POS Tags

Steve → NNP  
Jobs → NNP  
founded → VBD  
Apple → NNP

↓

NER

Steve Jobs → PERSON  
Apple → ORGANIZATION

# 4. Store Different Entity Types

We can also store detected entities in separate lists.

For example:

- People
- Organizations
- Locations

```python
text = """
Barack Obama visited Microsoft in the United States.
Bill Gates later travelled to Finland.
"""

tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)
tree = ne_chunk(pos_tags)

people = []
organizations = []
locations = []

for chunk in tree:

    if hasattr(chunk, 'label'):

        entity = " ".join(
            word for word, tag in chunk.leaves()
        )

        label = chunk.label()

        if label == "PERSON":
            people.append(entity)

        elif label == "ORGANIZATION":
            organizations.append(entity)

        elif label in ["GPE", "LOCATION"]:
            locations.append(entity)

print("People:")
print(people)

print("\nOrganizations:")
print(organizations)

print("\nLocations:")
print(locations)
```

# Why is NER Useful?

NER helps computers automatically extract important information from large amounts of text.

### Applications

- Extracting people's names from news articles
- Finding company names
- Finding countries and cities
- Extracting information from documents
- Search engines
- Question answering
- Information extraction
- Document analysis

### Simple Example

"John works for Google in London."

NER can transform this unstructured sentence into structured information:

| Entity | Type |
|---|---|
| John | PERSON |
| Google | ORGANIZATION |
| London | LOCATION |

# NER Summary

**Named Entity Recognition (NER)** identifies important entities in text.

### Common Entities

- PERSON → Person
- ORGANIZATION → Company or organization
- GPE / LOCATION → Country, city, or place
- DATE → Date
- TIME → Time
- MONEY → Monetary value

### Basic NLP Pipeline

**Text → Tokenization → POS Tagging → NER**

### Easy Way to Remember

> **POS tagging tells us what a word does grammatically.**
>
> **NER tells us what real-world entity a word or phrase represents.**

## Further NLP Topics

1. **Semantic Similarity**
   - Measuring how similar words, sentences, or documents are in meaning.

2. **Word Co-occurrence**
   - Studying which words frequently appear together.

3. **Bag of Words (BoW)**
   - Representing text using word occurrence or word frequency.

4. **TF-IDF**
   - Measuring how important a word is in a document compared with other documents.

5. **N-grams**
   - Analyzing sequences of words, such as bigrams and trigrams.

6. **Sentiment Analysis**
   - Identifying positive, negative, or neutral opinions in text.

7. **Text Classification**
   - Assigning text to predefined categories.

8. **Topic Modeling**
   - Discovering important topics from a collection of documents.

9. **Word Embeddings**
   - Representing words as numerical vectors.
   - Examples: Word2Vec, GloVe, FastText

10. **Transformers**
    - Modern deep-learning models for understanding language.
    - Examples: BERT, RoBERTa

11. **Large Language Models (LLMs)**
    - Advanced models capable of understanding and generating natural language.
