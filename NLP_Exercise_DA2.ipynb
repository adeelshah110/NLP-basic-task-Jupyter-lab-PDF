# NLP part 1 Tokenization
In NLP, tokenization means breaking text into smaller pieces called tokens. The main types are:
1. Word tokenization — splits text into words.

   | Type | Example |
|---|---|
| Word | `"I love Python"` → `["I", "love", "Python"]` |
| Sentence | `"Hi. How are you?"` → `["Hi.", "How are you?"]` |
| Character | `"cat"` → `["c", "a", "t"]` |
| Subword | `"playing"` → `["play", "ing"]` |

```python

text = "Python is a simple and powerful programming language for data processing."
# 1. Tokenization first way
tokens = word_tokenize(text)
print("Tokens:")
print(tokens)
```

# NLP Part 2 Stopwords  
Stopwords are very common words in a language that often carry little useful information for many NLP tasks.
Examples:
the, is, a, an, and, in, of, to, for

```python
stop_words = set(stopwords.words('english'))
stop_words_finnish = set(stopwords.words('swedish'))


text = "Python is a simple and powerful programming language for data processing."

#stop_words_finnish
```

```python
# 1. Tokenization first way
tokens = word_tokenize(text)
filtered_tokens = [
    word for word in tokens
    if word.lower() not in stop_words and word.isalpha()
]

print("\nAfter Stopword Removal:")
print(filtered_tokens)
```

# NLP part 3 Parts of Speech Taggings

```python
# 3. POS Tagging
pos_tags = pos_tag(filtered_tokens)

print("\nPOS Tags:")
print(pos_tags)
```

```python

# Example text
text = """
The intelligent student writes beautiful programs and analyzes large datasets.
She develops powerful applications and creates useful solutions.
The experienced teacher explains difficult concepts and gives practical examples.
"""

# 1. Tokenization
tokens = word_tokenize(text)

# 2. Remove stopwords and punctuation
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word.lower() not in stop_words and word.isalpha()
]

print("Filtered Tokens:")
print(filtered_tokens)

# 3. POS Tagging
pos_tags = pos_tag(filtered_tokens)

print("\nPOS Tags:")
print(pos_tags)

# 4. Separate Nouns, Verbs, and Adjectives

nouns = []
verbs = []
adjectives = []

for word, tag in pos_tags:

    # Nouns: NN, NNS, NNP, NNPS
    if tag.startswith('NN'):
        nouns.append(word)

    # Verbs: VB, VBD, VBG, VBN, VBP, VBZ
    elif tag.startswith('VB'):
        verbs.append(word)

    # Adjectives: JJ, JJR, JJS
    elif tag.startswith('JJ'):
        adjectives.append(word)

print("\nNouns:")
print(nouns)

print("\nVerbs:")
print(verbs)

print("\nAdjectives:")
print(adjectives)
```
